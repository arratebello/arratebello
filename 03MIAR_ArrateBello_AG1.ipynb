## Divide y vencerás - Problema Torres de Hanói
- Parto de un problema y lo soluciono mediante la solución de subproblemas y unión de resultados
- En este caso concreto: $H(n) = H(n-1) + 1 + H(n-1)$
- Pensar lo anterior como... "Muevo todos los discos salvo uno a torre auxiliar, muevo el disco más grande a la torre final y muevo todos los discos restantes de la torre auxiliar a la final"
- Solamente resuelvo el problema de mover un disco (condición de parada de división) porque el resto se aplica el mismo algoritmo.
- Mover un disco es muy sencillo -> Solución directa -> Mover de torre de incio a torre de fin
- Ojo, el print del "if" es mover un disco entre torres (cualesquiera) y el print del "else" es el caso de mover el disco grande a torre final

```python
def torres_hannoi(numero_discos, inicio, fin): 
    # inicio es torre 1 y fin es torre 3

    # print(f"torres_hannoi({numero_discos}, {inicio}, {fin})")

    # Si el número de discos es 1 -> Caso base, mover de inicio a fin
    if numero_discos == 1:
        # print(f"Caso base: Mover desde {inicio} a {fin} -- torres_hannoi({numero_discos}, {inicio}, {fin})")
        print(f"Mover desde {inicio} a {fin}")

    # Si el número de discos es mayor que 1 -> Mover resto de discos, mover disco único,
    # recuperar => Recursivamente
    else:
        torres_hannoi(numero_discos-1, inicio, 6-inicio-fin) 
        # 6-inicio-fin es la torre auxiliar. Cuando fin=1, la auxiliar es 2, y cuando fin=2, la auxiliar es 1. 
        # Si a 6=1+2+3 le quitamos al torre de inicio y la de fin, siempre vamos a obtener la torre auxiliar 
        # print(f"Caso central: Mover desde {inicio} a {fin} -- torres_hannoi({numero_discos}, {inicio}, {fin})")
        print(f"Mover desde {inicio} a {fin}")
        torres_hannoi(numero_discos-1, 6-inicio-fin, fin)

torres_hannoi(3,1,3)
```

## Técnicas voraces - Problema del cambio
- Para resolver el problema, elijo una solución "buena" sin pensar en consecuencias
- Sería como elegir algo "óptimo" localmente con la esperanza de que sea lo mejor en general
- No siempre funciona bien -> Si en el conjunto de las soluciones, dos están próximas entre sí y generan una solución mayor puede fallar (ej: monedas [1, 3, 4, 9])

Vamos a construir un algoritmo que nos calcule cuántas monedas de cada una de las monedas tenemos que dar para devolver un cambio determinado. 

Vamos a aplicar un algoritmo voraz, por lo tanto, vamos a devolver siempre una moneda del máximo valor posible. 

```python
def cambio_monedas(sistema, cambio):

    print(f"Sistema: {sistema}") # valores de las monedas, monedas disponibles
    print(f"Cambio: {cambio}") # lo que tenemos que devolver

    # Inicialización de resultados
    resultado = [0] * len(sistema) # guardamos cuántas monedas de cada tipo vamos a devolver

    # Para cada posible moneda, calculamos el número de monedas máximo que podemos dar
    # y lo restamos del cambio pendiente (mejor decisión local, que no tiene por qué
    # ser óptima
    for i in range(len(sistema)): # recorrer los tipos de monedas
        total_monedas = cambio//sistema[i] # División con resultado entero
        resultado[i] = total_monedas
        cambio = cambio - resultado[i]*sistema[i]

    return resultado

# Sistema óptimo
sistema = [25, 10, 5, 1]
cambio = 27
resultado = cambio_monedas(sistema, cambio)
print(f"Resultado: {resultado}")

print("")

# Sistema NO óptimo
sistema = [11, 5, 1]
cambio = 15
resultado = cambio_monedas(sistema, cambio)
print(f"Resultado: {resultado}")
```

En conclusión, tomar soluciones óptimas locales NO nos lleva siempre a encontrar una solución óptima global. 

Los algoritmos voraces nos permiten aproximarnos a una solución buena, sin ser la mejor, pero de una manera mucho más rápida. 

```python
# Otra forma (muy parecida)

def cambio_monedas2(sistema, cambio):

    print(f"Sistema: {sistema}") # valores de las monedas, monedas disponibles
    print(f"Cambio: {cambio}") # lo que tenemos que devolver

    valor_acumulado = 0

    # Inicialización de resultados
    resultado = [0] * len(sistema) # guardamos cuántas monedas de cada tipo vamos a devolver

    # Para cada posible moneda, calculamos el número de monedas máximo que podemos dar
    # y lo sumamos al valor_acumulado (mejor decisión local, que no tiene por qué ser óptima)
    for i, valor in enumerate(sistema):
        total_monedas = (cambio - valor_acumulado) // valor
        resultado[i] = total_monedas
        valor_acumulado = valor_acumulado + total_monedas*valor 

        if valor_acumulado == cambio:
            return resultado 

# Sistema óptimo
sistema = [25, 10, 5, 1]
cambio = 27
resultado = cambio_monedas2(sistema, cambio)
print(f"Resultado: {resultado}")

print("")

# Sistema NO óptimo
sistema = [11, 5, 1]
cambio = 15
resultado = cambio_monedas2(sistema, cambio)
print(f"Resultado: {resultado}")
```

## Backtracking - Problema de las N reinas
- Se divide el problema en etapas
- En cada etapa se construye una solución parcial
- Se comprueba si esa solución parcial sigue siendo válida
- Si es válida, se continúa con la siguiente etapa
- Si no es válida, se descarta y se prueba otra alternativa
- Si ninguna alternativa funciona, se retrocede a la etapa anterior para cambiar decisiones

El problema consiste en colocar $N$ reinas en un tablero $N \times N$ de tal manera que no se amenacen entre ellas. 

Solución: $N$-tuplas $(x_1,...,x_N)$ donde el valor de cada elemento es la posición (fila) de una reina en la columna i-ésima. 

```python
# Función que comprueba si por columnas hay un error en la solución
# etapa(i) = etapa(j) -> Error
# También comprueba si por diagonales hay un error en la solución
# (abs(i-j) = abs(xi-xj)) -> Error
def conflicto(solucion, etapa):
    for i in range(etapa+1): 
        if solucion.count(solucion[i]) > 1: # verifica si existen 2 reinas en la misma fila
            return True
        for j in range(etapa):
            if (i != j):
                if (abs(i-j) == abs(solucion[i]-solucion[j])): # verifica que no estén en la misma diagonal
                    return True
    else:
        return False
```

```python
# Búsqueda de una solución
def colocar_reinas(numero_reinas, solucion = None, etapa = 0):

    # Definir el vector de soluciones
    if solucion is None:
        solucion = numero_reinas*[0] # lista de longitud numero_reinas

    # Caso base: si he colocado todas las reinas (he completado todas las etapas)
    # etapa comienza en 0 y llega hasta numero_reinas
    if numero_reinas == etapa: # si ya hemos colocado todas las reinas
        #print(f"({numero_reinas}, {solucion}, {etapa})")
        print(f"{solucion}")
        return solucion

    else: # si estamos en alguna etapa anterior
        # Recorro todas las columnas posibles para la reina en la fila (etapa)
        for i in range(1,numero_reinas+1):
            solucion[etapa] = i
            #print(f"({numero_reinas}, {solucion}, {etapa})")

            # Compruebo si la colocación actual no entra en conflicto
            # con reinas anteriores (columnas y diagonales)
            if not (conflicto(solucion, etapa)):
                # Si es válida, intento resolver el resto del tablero
                resultado = colocar_reinas(numero_reinas, solucion, etapa + 1)

                # Si desde aquí se encuentra una solución válida,
                # la propago hacia arriba para detener la búsqueda
                # En otras palabras, ya tengo resultado, no necesito
                # seguir probando, lo devuelvo
                if resultado is not None:
                    return resultado

        # Limpio la posición de la reina si todos los estados fallan (vuelta atrás)
        solucion[etapa] = 0

        # Si ninguna columna funciona en esta etapa, esta rama no tiene solución
        #return None

solucion = colocar_reinas(4)
```

Problema: Encontrar los dos puntos más cercanos
- Dado un conjunto de puntos se trata de encontrar los dos puntos más cercanos

Guía para aprendizaje:
- Suponer en 1D, o sea, una lista de números: [3403, 4537, 9089, 9746, 7259, ….]
- Primer intento: Fuerza bruta
- Calcular la complejidad. ¿Se puede mejorar?
- Segundo intento. Aplicar Divide y Vencerás
- Calcular la complejidad. ¿Se puede mejorar?
- Extender el algoritmo a 2D: [(1122, 6175), (135, 4076), (7296, 2741)…]
- Extender el algoritmo a 3D.

#### 1. Intento: Fuerza bruta 

```python
def puntos_cercanos_fb(puntos):

    mejor_distancia = float('inf') # variable en la que guardaremos la distancia más cercana
    mejor_par = None # inicializamos los dos puntos más cercanos, al principio, ninguno

    # comparamos cada punto con todos los siguientes (todos los pares posibles)
    for i in range(len(puntos)):
        for j in range(i+1, len(puntos)):
            distancia = abs(puntos[i] - puntos[j])
            if distancia < mejor_distancia:
                mejor_distancia = distancia
                mejor_par = [puntos[i], puntos[j]]
    
    return mejor_par, mejor_distancia
```

```python
puntos = [3403, 4537, 9089, 9746, 7259, 4600]
mejor_par, mejor_distancia = puntos_cercanos_fb(puntos)
print(f"Los puntos más cercanos son {mejor_par} con una distancia de {mejor_distancia}.")
```

##### Complejidad

Como comparamos todos los pares de puntos posibles y tenemos dos for, la complejidad del algoritmo es de $O(n²)$. 


#### 2. Intento: Divide y vencerás

Para aplicar Divide y vencerás, el algoritmo va a seguir los siguientes pasos: 
- Ordenamos de menor a mayor los puntos primero 
- Caso base: Vamos a considerar el caso base cuando la lista de puntos tenga longitud 3 o menos. En ese caso, vamos a aplicar el algoritmo de fuerza bruta. 
- Cuando no estemos en el caso base, dividimos la lista en dos mitades y resolvemos cada mitad por separado, usando la recursividad. 
- Una vez que la recursividad nos devuelve el mejor par de la mitad izquierda y el de la mitad derecha, los comparamos entre sí. Nos quedamos con el par que tenga la distancia más pequeña.
- Por último, al haber partido la lista, tenemos que tener en cuenta que existe la posibilidad de que el verdadero par más cercano esté compuesto por el último punto del lado izquierdo y el primer punto del lado derecho. Comparamos estos dos puntos y si la distancia entre esos dos puntos centrales es menor que la mejord istancia de antes, actualizamos la variable con el par ganador.

```python
def recursividad_dyv(puntos_ordenados):

    # Caso base: si hay 3 o menos puntos, usamos la fuerza bruta
    if len(puntos_ordenados) <= 3:
        return puntos_cercanos_fb(puntos_ordenados)

    # dividimos la lista en dos mitades
    medio = len(puntos_ordenados) // 2
    
    # resolvemos cada mitad llamando a esta función (recursividad)
    par_izq, dist_izq = recursividad_dyv(puntos_ordenados[:medio])
    par_der, dist_der = recursividad_dyv(puntos_ordenados[medio:])

    # nos quedamos con la mejor distancia de las dos mitades
    if dist_izq < dist_der:
        mejor_par, mejor_distancia = par_izq, dist_izq
    else: 
        mejor_par, mejor_distancia = par_der, dist_der

    # comprobamos el par del medio 
    punto_izq = puntos_ordenados[medio - 1] # último de la izquierda
    punto_der = puntos_ordenados[medio] # primero de la derecha
    dist = abs(punto_izq - punto_der)
    
    if dist < mejor_distancia:  # comparamos si la distancia es menor que la que teníamos
        mejor_distancia = dist
        mejor_par = (punto_izq, punto_der)

    return mejor_par, mejor_distancia
```

```python
def puntos_cercanos_dyv(puntos):
    
    puntos_ordenados = sorted(puntos) # ordenamos la lista de puntos 

    # pasamos la lista ordenada a la función recursiva
    return recursividad_dyv(puntos_ordenados)
```

```python
puntos = [3403, 4537, 9089, 9746, 7259, 4600]
mejor_par, mejor_distancia = puntos_cercanos_dyv(puntos)
print(f"Los puntos más cercanos son {mejor_par} con una distancia de {mejor_distancia}.")
```

##### Complejidad

- La ordenación inicial con la función `sorted()` tiene un coste de $O(n \log n)$.  
- En cada paso, el algoritmo se llama a sí mismo dos veces con la mitad de los datos, lo que genera una relación de recurrencia de $2T(n/2)$.
- Dentro de esa recursividad, las operaciones de partición de la lista y la comprobación de los dos puntos centrales consumen un tiempo lineal de $O(n)$. Al agrupar estos factores, la ecuación de la recursividad es $T(n) = 2T(n/2) + O(n)$, que se resuelve como $O(n \log n)$. 
- Al coincidir con el tiempo de la ordenación inicial, el coste global del algoritmo es $O(n \log n)$.
